#Question 1
* Implement a program to convert a given crisp input into appropriate membership values using user-defined fuzzy membership functions.

```python
import math

def triangular_mf(x, a, b, c):
    if x <= a or x >= c:
        return 0.0
    elif a < x <= b:
        return (x - a) / (b - a)
    else:
        return (c - x) / (c - b)

def trapezoidal_mf(x, a, b, c, d):
    if x <= a or x >= d:
        return 0.0
    elif a < x < b:
        return (x - a) / (b - a)
    elif b <= x <= c:
        return 1.0
    else:
        return (d - x) / (d - c)

def gaussian_mf(x, mean, sigma):
    return math.exp(-((x - mean) ** 2) / (2 * sigma ** 2))

def sigmoidal_mf(x, a, c):
    return 1 / (1 + math.exp(-a * (x - c)))


x = float(input("Enter the crisp input value: "))

print("\nChoose a Membership Function:")
print("1. Triangular")
print("2. Trapezoidal")
print("3. Gaussian")
print("4. Sigmoidal")
choice = int(input("Enter your choice (1-4): "))

mf_name = None
membership = None

if choice == 1:
    a = float(input("Enter a (left base): "))
    b = float(input("Enter b (peak): "))
    c = float(input("Enter c (right base): "))
    membership = triangular_mf(x, a, b, c)
    mf_name = "Triangular"

elif choice == 2:
    a = float(input("Enter a (left base): "))
    b = float(input("Enter b (left top): "))
    c = float(input("Enter c (right top): "))
    d = float(input("Enter d (right base): "))
    membership = trapezoidal_mf(x, a, b, c, d)
    mf_name = "Trapezoidal"

elif choice == 3:
    mean = float(input("Enter mean: "))
    sigma = float(input("Enter sigma (spread): "))
    membership = gaussian_mf(x, mean, sigma)
    mf_name = "Gaussian"

elif choice == 4:
    a = float(input("Enter a (slope): "))
    c = float(input("Enter c (centre): "))
    membership = sigmoidal_mf(x, a, c)
    mf_name = "Sigmoidal"

else:
    print("Invalid choice!")

if membership is not None:
    print("\nOUTPUT\n")
    print("Crisp Input Value =", x)
    print("Membership Function Used =", mf_name)
    print("Membership Value (µ) =", round(membership, 4))
```

# Question 2
Implement all the defuzzification methods discussed in class:

- Max-Membership / Height Method
- Centroid Method
- Weighted Average Method
- Mean-Max Membership / Middle-of-Maxima Method
- Centre of Sums Method
- Centre of Highest / Largest Area Method
- First of Maxima
- Last of Maxima

A fuzzy output is generally obtained by combining (aggregating) the outputs of several fired
rules. So, the program below first takes a common Universe of Discourse, then takes the
membership values of **n** individual fuzzy output sets (rule outputs) defined over that
universe, and implements every defuzzification method as a separate user-defined function.

```python
def trapz(y, x):
    total = 0.0
    for i in range(len(x) - 1):
        total += (x[i + 1] - x[i]) * (y[i] + y[i + 1]) / 2
    return total

def height_method(x, mu):
    max_mu = max(mu)
    idx = mu.index(max_mu)
    return x[idx]

def centroid_method(x, mu):
    numerator = trapz([xi * mi for xi, mi in zip(x, mu)], x)
    denominator = trapz(mu, x)
    return numerator / denominator if denominator != 0 else 0.0

def weighted_average_method(sets):
    numerator = 0.0
    denominator = 0.0
    for (x, mu) in sets:
        h = max(mu)
        peak_x = x[mu.index(h)]
        numerator += peak_x * h
        denominator += h
    return numerator / denominator if denominator != 0 else 0.0

def mean_max_method(x, mu, tol=1e-9):
    max_mu = max(mu)
    xs = [xi for xi, mi in zip(x, mu) if abs(mi - max_mu) <= tol]
    return sum(xs) / len(xs)

def centre_of_sums_method(sets, x_universe):
    numerator = 0.0
    denominator = 0.0
    for (x, mu) in sets:
        area_i = trapz(mu, x)
        centroid_i = centroid_method(x, mu)
        numerator += centroid_i * area_i
        denominator += area_i
    return numerator / denominator if denominator != 0 else 0.0

def centre_of_largest_area_method(sets):
    best_area = -1
    best_centroid = 0.0
    for (x, mu) in sets:
        area_i = trapz(mu, x)
        if area_i > best_area:
            best_area = area_i
            best_centroid = centroid_method(x, mu)
    return best_centroid

def first_of_maxima(x, mu, tol=1e-9):
    max_mu = max(mu)
    for xi, mi in zip(x, mu):
        if abs(mi - max_mu) <= tol:
            return xi

def last_of_maxima(x, mu, tol=1e-9):
    max_mu = max(mu)
    for xi, mi in reversed(list(zip(x, mu))):
        if abs(mi - max_mu) <= tol:
            return xi

print("\nDefuzzification: Input Fuzzy Output Set(s)\n")

x_vals = list(map(float, input("Enter the common Universe of Discourse (comma separated x values): ").split(",")))

n = int(input("Enter number of fuzzy output sets (rule outputs) to combine: "))

sets = []
for i in range(n):
    print(f"\nFuzzy Set {i + 1}:")
    mu_vals = list(map(float, input(f"  Enter {len(x_vals)} membership values (comma separated, in [0,1]): ").split(",")))
    for m in mu_vals:
        if m < 0 or m > 1:
            raise ValueError("Membership values must lie in the interval [0, 1].")
    sets.append((x_vals, mu_vals))

mu_agg = [max(mu[i] for (_, mu) in sets) for i in range(len(x_vals))]

print("\nAGGREGATED FUZZY OUTPUT SET\n")
print("x :", x_vals)
print("µ :", [round(m, 4) for m in mu_agg])

print("\nIndividual Method Outputs\n")
print("1. Max-Membership / Height Method =", round(height_method(x_vals, mu_agg), 4))
print("2. Centroid Method =", round(centroid_method(x_vals, mu_agg), 4))
print("3. Weighted Average Method =", round(weighted_average_method(sets), 4))
print("4. Mean-Max Membership Method =", round(mean_max_method(x_vals, mu_agg), 4))
print("5. Centre of Sums Method =", round(centre_of_sums_method(sets, x_vals), 4))
print("6. Centre of Largest Area Method =", round(centre_of_largest_area_method(sets), 4))
print("7. First of Maxima =", round(first_of_maxima(x_vals, mu_agg), 4))
print("8. Last of Maxima =", round(last_of_maxima(x_vals, mu_agg), 4))
```
